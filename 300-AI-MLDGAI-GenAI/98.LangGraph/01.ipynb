```python
#Create a service client by name using the default session.
import json
import os
import sys

import boto3
import botocore
from langchain_aws import ChatBedrock
from langchain_core.output_parsers import StrOutputParser
from langchain_core.messages import SystemMessage, HumanMessage

module_path = ".."
sys.path.append(os.path.abspath(module_path))

bedrock_client = boto3.client('bedrock-runtime',region_name=os.environ.get("AWS_DEFAULT_REGION", None))
```

```python
# from langchain_openai.chat_models import ChatOpenAI

model_id = "amazon.nova-lite-v1:0"
model_kwargs =  { 
        "temperature": 0,
        "top_p": 1,
}

model = ChatBedrock(
    client=bedrock_client,
    model_id=model_id,
    model_kwargs=model_kwargs,
)

response = model.invoke("서울은")
print(response.content)
```

```python
from langchain_core.messages import HumanMessage

prompt = [HumanMessage("대한민국의 수도는 어디야?")]

response = model.invoke(prompt)
print(response.content)
```

```python
from pydantic import BaseModel


class AnswerWithJustification(BaseModel):
    """사용자의 질문에 대한 답변과 그 답변에 대한 근거입니다."""

    answer: str
    """사용자의 질문에 대한 답변"""
    justification: str
    """답변에 대한 근거"""


# llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
llm = model
structured_llm = llm.with_structured_output(AnswerWithJustification)

response = structured_llm.invoke(
    "벽돌 1파운드와 깃털 1파운드 중 어느 것이 더 무겁나요?"
)
print(response)
```

### imperative way

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import chain

# 구성 요소

template = ChatPromptTemplate.from_messages(
    [
        ("system", "당신은 사용자를 돕는 유용한 어시스턴트입니다."),
        ("human", "{question}"),
    ]
)

# model = ChatOpenAI(model="gpt-4o-mini")

# 구성 요소들을 하나의 함수로 결합
# @chain 데코레이터는 작성한 모든 함수에 동일한 Runnable 인터페이스를 추가합니다.


@chain
def chatbot(values):
    prompt = template.invoke(values)
    return model.invoke(prompt)


# 사용하기

response = chatbot.invoke({"question": "어떤 모델 제공업체가 LLM을 제공하나요?"})
print(response.content)
```

### Declarative way

```python
from langchain_core.prompts import ChatPromptTemplate

# 구성 요소

template = ChatPromptTemplate.from_messages(
    [
        ("system", "당신은 사용자를 돕는 유용한 어시스턴트입니다."),
        ("human", "{question}"),
    ]
)

# model = ChatOpenAI()

# | 연산자를 사용하여 구성 요소들을 결합

chatbot = template | model

# 사용하기

response = chatbot.invoke({"question": "어떤 모델 제공업체가 LLM을 제공하나요?"})
print(response.content)

# 스트리밍

for part in chatbot.stream({"question": "어떤 모델 제공업체가 LLM을 제공하나요?"}):
    print(part)
```
